```python
import nltk
from nltk.util import ngrams
```

```python
sentence = "Natural Language Processing for Text and Speech."
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
words = nltk.word_tokenize(sentence)
words
```

```python
list(ngrams(words, 1))
```

```python
type(('Natural',))
```

```python
list(ngrams(words, 2))
```

```python
trigrams = list(ngrams(words, 3))
trigrams
```
