```python
!pip install gensim nltk
import gensim
from gensim.models import Word2Vec
import nltk
from nltk.tokenize import word_tokenize
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
sentences = {
    "I deposited money in the bank.",
    "We sat on the riverbank and watched the boats.",
    "The pilot had to bank the plane of the hill",
    "The children sledded down the bank of the hill",
    "There was a bank on monitors in the control room.",
    "He made a bank shot to pocket the ball."
}
```

```python
tokenized_sentences = [word_tokenize(sentence.lower()) for sentence in sentences]
```

```python
tokenized_sentences
```

```python
model = Word2Vec(sentences=tokenized_sentences, vector_size=50, window=5, min_count=1, workers=4)
```

```python
model.wv['bank']
```

```python
model.wv['money']
```

```python
model.wv.most_similar('bank')
```

```python
model.wv.most_similar('bank', topn=5)
```

```python
model_2 = Word2Vec(sentences=tokenized_sentences, vector_size=50, window=10, min_count=2, workers=4)
```

```python
model_2.wv['bank']
```

```python
model_2.wv['pilot']
```

```python
model_2.wv.most_similar('bank')
```
